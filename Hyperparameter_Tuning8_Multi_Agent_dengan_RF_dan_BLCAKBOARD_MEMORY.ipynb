```python
# ============================================================
# MULTI-AGENT INTELLIGENT GRID WORLD (FINAL VERSION)
# ============================================================
# Features:
# - Multi-agent (random start each episode)
# - Large grid world
# - Supervised ML brain (Decision Tree)
# - Reinforcement Learning (Q-learning)
# - Anti-deadlock (short-term memory)
# - Meta-learning epsilon (adaptive exploration)
# - Blackboard collective memory
# - Live animation (Play)
# - Explicit colors, legend, reward heatmap
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from matplotlib.colors import ListedColormap
import matplotlib.patches as mpatches
from IPython.display import HTML
from collections import deque, defaultdict
from sklearn.tree import DecisionTreeClassifier

# ============================================================
# 1. WORLD CONFIGURATION
# ============================================================

GRID_SIZE = 18
N_DIRT = 30
N_BLOCK = 35
N_AGENTS = 3

EMPTY, DIRT, BLOCK = 0, 1, -1

ACTIONS = {
    0: (-1, 0),  # UP
    1: (1, 0),   # DOWN
    2: (0, -1),  # LEFT
    3: (0, 1)    # RIGHT
}

# ============================================================
# 2. COLOR DEFINITIONS
# ============================================================

WORLD_COLORS = {
    "BLOCK": "black",
    "EMPTY": "white",
    "DIRT": "limegreen"
}

AGENT_COLORS = [
    "red", "blue", "orange", "purple", "cyan", "magenta"
]

world_cmap = ListedColormap([
    WORLD_COLORS["BLOCK"],  # index 0
    WORLD_COLORS["EMPTY"],  # index 1
    WORLD_COLORS["DIRT"]    # index 2
])

# ============================================================
# 3. BLACKBOARD (COLLECTIVE MEMORY)
# ============================================================

BLACKBOARD = {
    "paths": set(),
    "dirt_locations": set()
}

# ============================================================
# 4. SUPERVISED BRAIN (SHARED)
# ============================================================

X_train, y_train = [], []

for _ in range(1200):
    obs = np.random.choice([EMPTY, DIRT, BLOCK], 4)
    if DIRT in obs:
        action = int(np.where(obs == DIRT)[0][0])
    else:
        safe = [i for i, v in enumerate(obs) if v != BLOCK]
        action = np.random.choice(safe) if safe else 0
    X_train.append(obs)
    y_train.append(action)

brain = DecisionTreeClassifier(max_depth=6)
brain.fit(X_train, y_train)

# ============================================================
# 5. WORLD GENERATOR
# ============================================================

def create_world():
    grid = np.zeros((GRID_SIZE, GRID_SIZE), dtype=int)

    for _ in range(N_DIRT):
        x, y = np.random.randint(0, GRID_SIZE, 2)
        grid[x, y] = DIRT

    for _ in range(N_BLOCK):
        x, y = np.random.randint(0, GRID_SIZE, 2)
        if grid[x, y] == EMPTY:
            grid[x, y] = BLOCK

    return grid

# ============================================================
# 6. AGENT CLASS
# ============================================================

class Agent:
    """
    Multi-agent with:
    - local perception
    - supervised brain
    - Q-learning
    - meta-learning epsilon
    - short-term memory (anti-deadlock)
    """

    def __init__(self, name):
        self.name = name
        self.Q = defaultdict(lambda: np.zeros(4))
        self.visit_memory = deque(maxlen=15)
        self.epsilon = 0.3
        self.reset()

    def reset(self):
        self.pos = [
            np.random.randint(0, GRID_SIZE),
            np.random.randint(0, GRID_SIZE)
        ]

    def perceive(self, grid):
        obs = []
        for dx, dy in ACTIONS.values():
            nx, ny = self.pos[0] + dx, self.pos[1] + dy
            if nx < 0 or ny < 0 or nx >= GRID_SIZE or ny >= GRID_SIZE:
                obs.append(BLOCK)
            else:
                obs.append(grid[nx, ny])
        return np.array(obs)

    def step(self, grid, reward_heatmap):
        obs = self.perceive(grid)
        state = tuple(self.pos + obs.tolist())
        self.visit_memory.append(state)

        deadlock = self.visit_memory.count(state) > 2

        # ---------------- ACTION SELECTION ----------------
        if deadlock or np.random.rand() < self.epsilon:
            action = np.random.choice(list(ACTIONS.keys()))
        else:
            action = int(brain.predict([obs])[0])

        dx, dy = ACTIONS[action]
        nx, ny = self.pos[0] + dx, self.pos[1] + dy

        reward = -0.01
        if nx < 0 or ny < 0 or nx >= GRID_SIZE or ny >= GRID_SIZE:
            nx, ny = self.pos
            reward = -0.2
        elif grid[nx, ny] == BLOCK:
            nx, ny = self.pos
            reward = -0.2
        elif grid[nx, ny] == DIRT:
            reward = 1.0
            grid[nx, ny] = EMPTY
            BLACKBOARD["dirt_locations"].add((nx, ny))
            reward_heatmap[nx, ny] += 1

        # ---------------- RL UPDATE ----------------
        next_obs = self.perceive(grid)
        next_state = tuple([nx, ny] + next_obs.tolist())

        self.Q[state][action] += 0.1 * (
            reward + 0.9 * np.max(self.Q[next_state]) - self.Q[state][action]
        )

        # ---------------- META-LEARNING EPSILON ----------------
        if reward > 0:
            self.epsilon = max(0.05, self.epsilon * 0.99)
        else:
            self.epsilon = min(0.5, self.epsilon / 0.99)

        self.pos = [nx, ny]
        BLACKBOARD["paths"].add((nx, ny))

# ============================================================
# 7. INITIALIZE WORLD & AGENTS
# ============================================================

grid = create_world()
reward_heatmap = np.zeros((GRID_SIZE, GRID_SIZE))
agents = [Agent(f"A{i}") for i in range(N_AGENTS)]

# ============================================================
# 8. VISUALIZATION SETUP
# ============================================================

fig, ax = plt.subplots(figsize=(6,6))

im_world = ax.imshow(grid + 1, cmap=world_cmap, vmin=0, vmax=2)
im_heat = ax.imshow(reward_heatmap, cmap="hot", alpha=0.35, vmin=0)

agent_scatters = []
for i, agent in enumerate(agents):
    sc = ax.scatter(
        agent.pos[1],
        agent.pos[0],
        c=AGENT_COLORS[i % len(AGENT_COLORS)],
        s=80,
        edgecolors="black"
    )
    agent_scatters.append(sc)

# ============================================================
# 9. LEGEND (AUTOMATIC)
# ============================================================

legend_elements = [
    mpatches.Patch(color=WORLD_COLORS["EMPTY"], label="Empty"),
    mpatches.Patch(color=WORLD_COLORS["DIRT"], label="Dirt"),
    mpatches.Patch(color=WORLD_COLORS["BLOCK"], label="Block"),
]

for i, agent in enumerate(agents):
    legend_elements.append(
        mpatches.Patch(
            color=AGENT_COLORS[i % len(AGENT_COLORS)],
            label=f"Agent {agent.name}"
        )
    )

ax.legend(
    handles=legend_elements,
    loc="upper right",
    fontsize=8,
    framealpha=0.9
)

# ============================================================
# 10. ANIMATION LOOP
# ============================================================

def update(frame):
    global grid, reward_heatmap

    # Reset episode
    if frame % 150 == 0 and frame > 0:
        grid = create_world()
        reward_heatmap[:] = 0
        BLACKBOARD["paths"].clear()
        BLACKBOARD["dirt_locations"].clear()
        for agent in agents:
            agent.reset()

    # Agent actions
    for agent in agents:
        agent.step(grid, reward_heatmap)

    # Update visuals
    im_world.set_data(grid + 1)
    im_heat.set_data(reward_heatmap)

    for i, agent in enumerate(agents):
        agent_scatters[i].set_offsets([agent.pos[1], agent.pos[0]])

    ax.set_title(f"Multi-Agent Grid World | Step {frame}")
    return [im_world, im_heat, *agent_scatters]

ani = animation.FuncAnimation(
    fig, update, frames=500, interval=200
)

plt.close()
HTML(ani.to_jshtml())
```
